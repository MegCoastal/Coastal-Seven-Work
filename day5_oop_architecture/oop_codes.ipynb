```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def display(self):
        print(self.name)
        print(self.age)

s1 = Student('Ravi',20)
s2 = Student('Chandra',21)
s1.display()
s2.display()
```

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts")


c = Car()

c.start()
```
