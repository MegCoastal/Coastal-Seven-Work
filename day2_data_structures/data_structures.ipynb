```python
list = [{'name':'shyam','dept':'IT'},{'name' : 'ram', 'dept':'core'},{'name' : 'john','dept':'IT'}]
grp = {}
for emp in list :
    dept = emp['dept']
    grp.setdefault(dept,[]).append(emp['name'])

print(grp)
```
